```python
import numpy as np
import pandas as pd

from google.colab import files
uploaded = files.upload()

df = pd.read_csv('ObesityDataSet_raw_and_data_sinthetic.csv')

# Convert BMI column to a NumPy array
bmi_values = df['BMI'].values

# Replace missing BMI values with the mean BMI
mean_bmi = np.nanmean(bmi_values)
bmi_values = np.where(np.isnan(bmi_values), mean_bmi, bmi_values)

# Assign the cleaned values back to the DataFrame
df['BMI'] = bmi_values

# Convert BMI column to NumPy array
bmi_values = df['BMI'].values

# Keep only realistic BMI values
valid_bmi = bmi_values[(bmi_values >= 10) & (bmi_values <= 80)]

# Filter the DataFrame
df = df[df['BMI'].isin(valid_bmi)]

# Convert columns into NumPy arrays
bmi_values = df['BMI'].values
weight_values = df['Weight (KG)'].values
age_values = df['Age'].values

# Calculate descriptive statistics
avg_bmi = np.mean(bmi_values)
max_bmi = np.max(bmi_values)
avg_weight = np.mean(weight_values)
avg_age = np.mean(age_values)

print(f"Average BMI: {avg_bmi:.2f}")
print(f"Highest BMI: {max_bmi:.2f}")
print(f"Average Weight: {avg_weight:.2f} kg")
print(f"Average Age: {avg_age:.2f} years")

# Convert Weight column to NumPy array
weights = df['Weight (KG)'].values

# Min-Max scaling
min_weight = np.min(weights)
max_weight = np.max(weights)

normalized_weight = (weights - min_weight) / (max_weight - min_weight)

# Add normalized column back to DataFrame
df['Normalized_Weight'] = normalized_weight

# Convert numerical columns to consistent numeric formats
df['Age'] = df['Age'].astype(np.int64)
df['Weight (KG)'] = df['Weight (KG)'].astype(float)
df['BMI'] = df['BMI'].astype(float)
df['Water Intake'] = df['Water Intake'].astype(float)
df['Physical Activity'] = df['Physical Activity'].astype(float)
df['Technology Use'] = df['Technology Use'].astype(float)

from sklearn.preprocessing import MinMaxScaler

# Select numerical columns to normalize
numeric_cols = ['Age', 'Heigh (M)t', 'Weight (KG)', 'BMI']

# Apply Min-Max Scaling
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print(df[numeric_cols].head())

# Create summary table
summary_table = df.describe()

# Show the summary table
summary_table

summary_table.to_csv("summary_table.csv")

# Check missing values in each column
missing_values = df.isnull().sum()
print("Missing Values:")
print(missing_values)

# Check data types
print("Data Types:")
print(df.dtypes)

# Check duplicate rows
duplicate_count = df.duplicated().sum()
print("Duplicate Rows:", duplicate_count)

# Remove duplicates if any exist
df = df.drop_duplicates()

# Summary statistics to check min, max, mean, and standard deviation
summary_stats = df.describe()
print(summary_stats)

# Check unique values in categorical columns
for col in df.select_dtypes(include=['object']).columns:
    print(f"\nUnique values in {col}:")
    print(df[col].unique())

# Remove extra spaces from text columns
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# Standardize yes/no columns if needed
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

def data_quality_report(df):
    print("Missing Values:\n", df.isnull().sum(), "\n")
    print("Data Types:\n", df.dtypes, "\n")
    print("Duplicate Rows:", df.duplicated().sum(), "\n")
    print("Summary Statistics:\n", df.describe(), "\n")

    for col in df.select_dtypes(include=['object']).columns:
        print(f"Unique values in {col}:\n", df[col].unique(), "\n")

data_quality_report(df)

df.to_csv("cleaned_obesity_dataset.csv", index=False)
```
