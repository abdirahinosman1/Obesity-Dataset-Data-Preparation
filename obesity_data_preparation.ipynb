```python
import numpy as np
import pandas as pd
from google.colab import files
from sklearn.preprocessing import MinMaxScaler

# Upload the CSV file
uploaded = files.upload()

# Load dataset
df = pd.read_csv('ObesityDataSet_raw_and_data_sinthetic.csv')

# Preview dataset
print("First five rows:")
display(df.head())

# --------------------------------------------------
# Step 1: Clean column formats and text formatting
# --------------------------------------------------

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Remove extra spaces from text/categorical columns
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# --------------------------------------------------
# Step 2: Convert numerical columns to correct formats
# --------------------------------------------------

numeric_cols = [
    'Age',
    'Heigh (M)t',
    'Weight (KG)',
    'BMI',
    'Vegetable Consumption',
    'Number of Main Meals',
    'Water Intake',
    'Physical Activity',
    'Technology Use'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert Age to integer after handling missing values later
# Do not convert yet if there are missing values

# --------------------------------------------------
# Step 3: Handle missing values
# --------------------------------------------------

# Replace missing numerical values with the mean of each column
for col in numeric_cols:
    mean_value = np.nanmean(df[col].values)
    df[col] = np.where(np.isnan(df[col]), mean_value, df[col])

# Convert Age to integer after missing values are handled
df['Age'] = df['Age'].round().astype(np.int64)

# --------------------------------------------------
# Step 4: Filter unrealistic outliers
# --------------------------------------------------

# Keep only realistic BMI values
df = df[(df['BMI'] >= 10) & (df['BMI'] <= 80)]

# Keep only realistic Age values
df = df[(df['Age'] >= 0) & (df['Age'] <= 100)]

# Keep only realistic Height values in meters
df = df[(df['Heigh (M)t'] >= 1.0) & (df['Heigh (M)t'] <= 2.5)]

# Keep only realistic Weight values in KG
df = df[(df['Weight (KG)'] >= 20) & (df['Weight (KG)'] <= 250)]

# --------------------------------------------------
# Step 5: Remove duplicate rows
# --------------------------------------------------

duplicate_count = df.duplicated().sum()
print("Duplicate Rows Before Cleaning:", duplicate_count)

df = df.drop_duplicates()

print("Duplicate Rows After Cleaning:", df.duplicated().sum())

# --------------------------------------------------
# Step 6: Calculate descriptive statistics
# --------------------------------------------------

bmi_values = df['BMI'].values
weight_values = df['Weight (KG)'].values
age_values = df['Age'].values

avg_bmi = np.mean(bmi_values)
max_bmi = np.max(bmi_values)
avg_weight = np.mean(weight_values)
avg_age = np.mean(age_values)

print(f"Average BMI: {avg_bmi:.2f}")
print(f"Highest BMI: {max_bmi:.2f}")
print(f"Average Weight: {avg_weight:.2f} kg")
print(f"Average Age: {avg_age:.2f} years")

# --------------------------------------------------
# Step 7: Normalize data WITHOUT replacing original columns
# --------------------------------------------------

cols_to_normalize = ['Age', 'Heigh (M)t', 'Weight (KG)', 'BMI']

scaler = MinMaxScaler()
normalized_values = scaler.fit_transform(df[cols_to_normalize])

df[['Normalized_Age',
    'Normalized_Height',
    'Normalized_Weight',
    'Normalized_BMI']] = normalized_values

print("Normalized columns preview:")
display(df[['Age', 'Heigh (M)t', 'Weight (KG)', 'BMI',
            'Normalized_Age', 'Normalized_Height',
            'Normalized_Weight', 'Normalized_BMI']].head())

# --------------------------------------------------
# Step 8: Create summary table
# --------------------------------------------------

summary_table = df.describe()

print("Summary Table:")
display(summary_table)

# Save summary table
summary_table.to_csv("summary_table.csv")

# --------------------------------------------------
# Step 9: Data quality checks
# --------------------------------------------------

print("Missing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nSummary Statistics:")
display(df.describe())

print("\nUnique Values in Categorical Columns:")
for col in df.select_dtypes(include=['object']).columns:
    print(f"\nUnique values in {col}:")
    print(df[col].unique())

# --------------------------------------------------
# Step 10: Automated data quality report
# --------------------------------------------------

def data_quality_report(df):
    print("Missing Values:\n", df.isnull().sum(), "\n")
    print("Data Types:\n", df.dtypes, "\n")
    print("Duplicate Rows:", df.duplicated().sum(), "\n")
    print("Summary Statistics:\n")
    display(df.describe())

    for col in df.select_dtypes(include=['object']).columns:
        print(f"Unique values in {col}:\n", df[col].unique(), "\n")

data_quality_report(df)

# --------------------------------------------------
# Step 11: Save cleaned dataset
# --------------------------------------------------

df.to_csv("cleaned_obesity_dataset.csv", index=False)

print("Files saved successfully:")
print("cleaned_obesity_dataset.csv")
print("summary_table.csv")

# Download files
files.download("cleaned_obesity_dataset.csv")
files.download("summary_table.csv")
```
